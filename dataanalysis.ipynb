```python
# find highest and lowest sales based on locaion
# represent the sales on a bar chart,also show the market share for each location using pie chart
```

```python
location=data.groupby('Location')
```

```python
location_data=[i for i,j in location]
```

```python
location_data
```

```python
data.groupby('Location').sum()['Total']
```

```python
import matplotlib.pyplot as plt
```

```python
plt.bar(location_data,data.groupby('Location').sum()['Total'])
```

```python
plt.pie(data.groupby('Location').sum()['Total'],labels=location_data,autopct='%1.1f%%')
```
