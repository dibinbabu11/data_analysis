```python
car=pd.read_csv('car-sales.csv')
```

```python
car   
```

```python
car['Price'] = car['Price'].str.replace(r'[$,.]', '', regex=True)  # Remove $, commas, and dots
car['Price'] = car['Price'].str[:-2]  # Remove last two characters (if needed)
car['Price'] = pd.to_numeric(car['Price'], errors='coerce')  # Convert to numeric, handle errors
car['sales date'] = pd.date_range('1/1/2024', periods=len(car), freq='D')  # Generate sequential dates
car['total'] = car['Price'].astype(int).cumsum()  # Cumulative sum of prices
car
```

```python
car.plot(x='sales date',y='total')
```
