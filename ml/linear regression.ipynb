```python
#x=input/independent,
#y=target
#x=age
#y=salary
```

```python
x=data['Age']
y=data['Salary']
```

```python
#reshape
x=x.values.reshape(-1,1)
```

```python
x
```

```python
x.shape
```

```python
#splitting data into train data and test data
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25)
#.25=25 perc of test data,75perc of train data  (entire data set is divided into two part)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
reg=LinearRegression()
```

```python
reg.fit(x_train,y_train)   #trining the data using train data
```

```python
#coefficent,intercpet
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
y_pred=reg.predict(x_test)
```

```python
y_pred
```

```python
#calculate mean squre error
from sklearn.metrics import mean_squared_error
```

```python
MSE=mean_squared_error(y_pred,y_test)
```

```python
MSE
```

```python
sum((y_test-y_pred)**2/len(y_pred))  #manually calculating msc
```

```python
#creating a dataframe to view result
```

```python
result=pd.DataFrame(columns=['ID','Age','Actual','Predicted'])
result['Actual']=y_test
result['Predicted']=y_pred
result['Age']=x_test

result=result.reset_index()
result['ID']=result.index
result.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```
