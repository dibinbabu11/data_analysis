# Regularization is a technique used to reduce overfitting by adding a penalty to the model's loss function.
# It helps the model generalize better on unseen data.

```python
data=pd.read_csv('Advertising.csv')
```

```python
data.head()
```

```python
data.isna().sum()
```

```python
data.duplicated().sum()
```

```python
x=data.drop(['sales'],axis=1)
```

```python
y=data['sales']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr=LinearRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
y_pred=lr.predict(x_test)
```

```python
y_pred
```

```python
from sklearn.metrics import mean_squared_error
```

```python
#testing error
testing_msc=mean_squared_error(y_pred,y_test)
testing_msc
```

```python
#training error
y_pred_train=lr.predict(x_train)

training_msc=mean_squared_error(y_pred_train,y_train)
training_msc


```
